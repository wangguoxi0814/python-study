# 异步上下文管理器

## 方式一: 异步上下文管理器类
- 创建一个类，实现__aenter__和__aexit__方法，这2个方法必须是协程方法
- 使用：`async with AsyncContextManager() as obj`

### 创建异步上下文管理器类

```python
class AsyncContextManager:
    async def __aenter__(self):
        print('__aenter__')
        return self

    async def __aexit__(self, exc_type, exc_val, exc_tb):
        print('__aexit__')
```

### 使用

```python
async with AsyncContextManager() as obj:
    print(obj)
    print('...')
```

## 方式二: @asynccontextmanager
- 使用@asynccontextmanager装饰一个协程函数，返回一个异步上下文管理器类
- 通过yield划分异步上下文的进入和退出
- 使用：`async with context_manager() as obj`

#### 使用

```python
from contextlib import asynccontextmanager


@asynccontextmanager
async def context_manager():
    print('__aenter__')
    yield 1
    print('__aexit__')

class MyService:

    @asynccontextmanager
    async def class_context_manager(self):
        print('__aenter__')
        yield self
        print('__aexit__')
```

```python
async with context_manager() as cm:
    print(cm)
```

```python
obj = MyService()
async with obj.class_context_manager() as cm:
    print(cm)
```

## 意义和用途
在进入和退出时，可以使用异步操作，普通的上下文管理器无法做到
- 适用场景
    - 异步数据库连接管理
    - 异步HTTP会话管理

```python
async def connect_db():
    pass

async def close_db(db):
    pass

@asynccontextmanager
async def get_db():
    db = await connect_db()
    yield db
    await close_db(db)

async with get_db() as db:
    print(db)
```
