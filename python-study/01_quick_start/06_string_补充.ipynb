# String

## replace
- 替换字符串，不改变原字符串,返回更换后的字符串

```python
s1 = 'Hello World$$'
print(s1.replace('$', '*'))
print(s1)
```

## split
- 按指定字符分割字符串，返回列表

```python
s2 = 'Hello*Python*World'
print(s2.split('*'))
print(s2)
```

## format
- 按位置参数格式化
- 按关键字参数格式化

```python
s3 = 'Hello {0} {1}'.format('Python', 'World')
print(s3)

s4 = '你好, {content}, {say}'.format(content='祖国', say='欢迎光临')
print(s4)
```
