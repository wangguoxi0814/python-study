# 内置函数

## hasattr() 和getattr()
操作对象可以是类、对象、模块、函数等
- hasattr(obj, attr): 判断对象是否包含属性
- getattr(obj, attr, default): 获取对象的属性值，可设置默认值。如果没有设置默认值，获取不存在的attr报错


### 操作类和对象的属性

```python
class Person():

    domain: str = '中国'

    def __init__(self, name, age):
        self.name = name
        self.age = age

p1 = Person('Peter', 22)
if hasattr(p1, 'name'):
    print(getattr(p1, 'name'))

# 获取不存在的属性报错，可设置默认值
print(getattr(p1, 'gender', '男'))

print(getattr(Person, '__name__'))
if hasattr(Person, 'domain'):
    print('存在domain')
    print(getattr(Person, 'domain'))
```

### 函数属性

```python
def work():
    print('working...')

work.func = '记录工作'

print(getattr(work, '__name__'))

if hasattr(work, 'func'):
    print('存在func属性')
    print(getattr(work, 'func'))
```

```python
import attr_test_module

print(getattr(attr_test_module, '__name__'))
print(getattr(attr_test_module, '__file__'))
print(getattr(attr_test_module, '__package__'))

if hasattr(attr_test_module, 'domain'):
    print('存在domain属性')
    print(getattr(attr_test_module, 'domain'))
else:
    print('不存在domain属性')
```
