# 类型注解
- python3.5+类型注解,`from typing import List, Dict, Tuple等`
- python3.9+类型注解直接小写: list[int], dict[str, str], tuple[int],更简洁

## 变量类型注解（python3.9+ 推荐）

```python
# 变量类型注解，限制变量的类型，或者容器内部元素的类型

s: str = "Hello"
print(s)
s = 1  # 警告
print(s)

# 容器, list, set同样写法
l: list[int] = [1, 2, 3]
ls: list[int | str] = [1, 'Hello']
# python可以推断出变量类型，这里append字符串会给警告，但如果对l1重新赋值，则不会警告，如果手动声明了类型注解，会警告重新赋值的操作
l1 = [1, 2, 3]

# dict
d: dict[str, int] = {'US': 90, 'CN': 99}
m: dict[str | int, int] = {'US': 90, 'CN': 99, 100: 100}

# tuple tuple的类型注解要和元素数量一致
# 指定个数
t: tuple[int] = (1,)
t1: tuple[int, int, int] = (1, 2, 3)
# 多个
t2: tuple[int, ...] = (1, 2, 3, 4)
t3: tuple[int | str, ...] = (1, 2, "Hello", 3)
```

## typing类型注解模块(python3.5+)

### List

```python
from typing import List

l: List[int] = [1, 2, 3]
print(l)
```

### Callable
- Callable[..., Any]: 参数任意类型，返回任意类型
- Callable[[int, str], Any]: 参数int, str，返回任意类型
- Callable[[], bool]: 无参数，返回bool

```python

from typing import Callable

def coding() -> str:
    return '写代码'

def clock(name: str, action: str) -> str:
    return f'{name} {action}打卡成功!'

def meet(type: str):
    return f'参加{type}会议'

def work(coding: Callable[[], str], clock: Callable[[str, str], str], meet: Callable[[str], str]):
    print(clock('Peter', '上班'))
    print(meet("需求评审"))
    print(f'工作：{coding()}')
    print(clock('Peter', '下班'))



work(coding, clock, meet)
```

## Annotated
    - 自定义元数据
    - 可通过`get_type_hints`和`__metadata__`实现自定义元数据处理逻辑，这也是Annotated在不同框架中都有不一样的写法.Annotated可以理解为值是一个容器
    - 在各个框架用法：
        - LangGraph: Annotated(list[str], add).在节点执行完修改状态时，通过add合并各个节点的list
        - FastAPI: 通过Annotated实现参数验证
        - Pydantic: 通过Annotated实现参数验证

```python
from typing import Annotated, get_type_hints


class AnnotatedDemo:
    name: Annotated[str, '姓名']
    age: Annotated[int, '年龄']

hints = get_type_hints(AnnotatedDemo, include_extras=True)
print(hints)
for k, v in hints.items():
    print(v.__metadata__)
```

## TypedDict
- 声明固定键值类型的字典,Python3.11版本前无法实例化
- 属性调用`['属性名']`
    - 好处：
        - 编辑器可以更好的校验字段
        - 可以结合Annotated给字段添加元数据和限制

```python
from typing import TypedDict, Annotated


class PersonalInfo(TypedDict):
    name: str
    age: int
    hobbies: Annotated[list[str], '兴趣爱好']

perter_info: PersonalInfo = {
    "name": "Peter",
    "age": 25,
    "hobbies": ['bk','swim']
}

print(perter_info)
print(perter_info['name'])
```
