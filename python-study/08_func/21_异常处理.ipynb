## 异常处理
- try:                    尝试捕获代码块
- except:                 捕获异常
- else:                   没有异常执行
- finally:                无论有没有异常都执行

### try except捕获异常最佳事件
- raise抛出异常
- traceback.format_exc():最灵活，拿到了完整字符串，方便处理
- logggig.exception(): 最推荐，自动记录完成对战信息

```python
try:
    a = 1 / 0
except Exception as e:
    print(f'e: {e}')
    print(f'e.args: {e.args}')
    print(f'异常文件: {e.__traceback__.tb_frame.f_code.co_filename}')
    print(f'异常行数: {e.__traceback__.tb_lineno}')
    import traceback
    print(f'format_exc: {traceback.format_exc()}')
    raise e
else:
    print('success')
finally:
    print('close resource')
```

```python
import logging

try:
    a = 1 / 0
except Exception as e:
    logging.exception('出错了！')
    raise e
else:
    print('success')
finally:
    print('close resource')
```

## 自定义异常
- 自定义异常，继承Exception或其子类，名字以Error结尾，如没有特殊要求，用pass占位即可

```python
class EmailFormatError(Exception):
    pass

raise EmailFormatError('邮箱格式错误!')
```

```python
# 自定义异常，加固定前缀
class IdcardError(Exception):

    def __init__(self, msg):
        super().__init__('【自定义异常】' + msg)

raise IdcardError('身份证识别异常!')
```
