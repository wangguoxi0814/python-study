## 定义枚举类
1. 继承Enum
2. 格式：`Name=Value`，左边恒为`name`，右边恒为`value`
3. 如果有多个属性，value应当为一个元组，且需要自定义`__init__`，参数要和元组元素一一对应，底层会遍历元组，一一设置到属性上。如果value是list，则会把list整体当做一个属性

```python
from enum import Enum


class HttpStatus(Enum):
    OK = (200, "success", "成功!")
    BAD_REQUEST = (400, "Bad Request", "请求错误！")
    SERVER_ERROR = (500, "Server Error", "服务端错误!")

    def __init__(self, code, text, description):
        self.code = code
        self.text = text
        self.description = description


print(HttpStatus.OK)
print(HttpStatus.BAD_REQUEST.name)
print(HttpStatus.BAD_REQUEST.value)
print(type(HttpStatus.BAD_REQUEST.value))  # tuple
print(HttpStatus.BAD_REQUEST.text)
print(HttpStatus.BAD_REQUEST.description)
print(HttpStatus.SERVER_ERROR)

print()
```

## @staticmethod和@classmethod
1. 区别：
    - @staticmethod是静态方法，入参没有`cls`
    - @staticmethod适合做工具方法、校验等
    - @classmethod是类方法，第一个入参是`cls`
    - @classmethod适合做工厂方法
2. 都能够被子类继承

## 对比Java和Python的import
- Java的导入是编译期行为，只是为了书写简单，不导也可以，但需要写全限定名称
- Python的导入是运行时行为，它会把模块加载到内存，当前作用域中，不import就无法使用包的功能
- Java会有全局包名冲突，这也是为什么Java的目录层级都遵循反向域名规范的原因，利用了DNS域名全球唯一性，保证不冲突
- Python的导入不会包全局包名冲突，因为PyPI限制了包名必须唯一。而模块名会冲突
    - 工作区和第三方包冲突：如果当前工作区有模块要json,而像使用第三方json，导入`import json`会导入工作区的json
    - 不同的包可能存在相同模块名，后引入的会覆盖前者，但可以通过`as`别名解决
