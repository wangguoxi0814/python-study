## time和datetime
- time底层是C语言，可以打印时间戳（自1970年01月01日 00:00:00 UTC以来的秒数）,适合性能计时，休眠
- datetime:是高级日期时间处理，面向对象设计。包括：`date`、`time`、`datetime`、`timedelta`

datetime以及相关类关键点：
- date: 日期
- time: 时间
- datetime: 日期+时间
- timedelta: 时间段
- strftime(pattern): 全名是string format time, 是将时间格式化为字符串，需要指定格式化格式。datetime\date\time都支持。格式化后不会影响原始对象。
- strptime(string, pattern)全名是 string parse time, 是将字符串解析为时间，需要描述被解析字符串的格式。只有datetime支持。如果解析后需要获取日期或时间，需要通过`.date()`或者`.time()`获取
- 时间格式：%Y-%m-%d %H:%M:%S
- 时间格式不同于Java\JavaScript: yyyy-MM-dd HH:mm:ss

### date

```python
from datetime import datetime, date

# ========== 1. date 类（日期：年-月-日） ==========

# 创建日期
d1 = date(2026, 8, 5)           # 指定日期
d2 = date.today()                 # 今天
d3 = date.fromtimestamp(1700000000)  # 从时间戳创建
d4 = date.fromisoformat("2024-01-15")  # 从ISO字符串创建（ISO标准：日期用-连接，时间用：连接，日期和时间中用T连接，比如：2025-06-07T19:09:10）

print(f"指定日期: {d1}")
print(f"今天: {d2}")
print(f"时间戳转日期: {d3}")
print(f"ISO字符串转日期: {d4}")

# 访问属性
print(f"\n年: {d1.year}, 月: {d1.month}, 日: {d1.day}")
print(f"星期几(0=周一): {d1.weekday()}")
print(f"星期几(1=周一): {d1.isoweekday()}")
# timetupe()讲date转为一个struct_time元组
print(f"d1.timetuple(): {d1.timetuple()}")
print(f"一年中第几天: {d1.timetuple().tm_yday}")

# 格式化，strftime将date转为字符串
print(f"\n格式化: {d1.strftime('%Y年%m月%d日')}")
# 把日期转化为ISO标准字符串
print(f"ISO格式: {d1.isoformat()}")
# 注意：date类没有strptime方法！只有datetime类有
# 正确做法：先用datetime解析，再取.date()
# strptime将字符串解析为datetime
print(f"字符串解析: {datetime.strptime('2024-01-15', '%Y-%m-%d').date()}")

# 日期比较
print(f"\nd1 < d2: {d1 < d2}")
print(f"d1 == d2: {d1 == d2}")

diff = d1 - d3
print(f'diff: {diff}')
print(f'diff type: {type(diff)}')
```

### time(datetime.time)

```python
from datetime import datetime, time, timedelta

# ========== 2. time 类（时间：时:分:秒.微秒） ==========

# 创建时间
t1 = time(14, 30, 0)             # 指定时分秒
t2 = time(14, 30, 45, 123456)    # 含微秒
t3 = time.fromisoformat("14:30:00")  # 从ISO字符串创建

print(f"时分秒: {t1}")
print(f"含微秒: {t2}")
print(f"ISO解析: {t3}")

# 访问属性
print(f"\n时: {t2.hour}, 分: {t2.minute}, 秒: {t2.second}, 微秒: {t2.microsecond}")

# 格式化
print(f"\n格式化: {t1.strftime('%H时%M分%S秒')}")
print(f"ISO格式: {t1.isoformat()}")
# 解析
print(f"解析：{datetime.strptime('08:44:10', '%H:%M:%S').time()}")

# 注意：time类不能做算术运算！只能比较大小
print(f"\nt1 < t2: {t1 < t2}")
```

### datetime
- datetime.now(): 获取本地时间，时区跟随操作系统
- datetime.now(timezone): 获取指定时区时间

```python
from datetime import datetime, time, timedelta

# ========== 3. datetime 类（日期 + 时间，最常用） ==========

# 创建
dt1 = datetime(2024, 1, 15, 14, 30, 0, 111)   # 指定完整时间
dt2 = datetime.now()                       # 当前本地时间
dt3 = datetime.utcnow()                    # 当前UTC时间（已弃用，建议用下面的方式）
dt4 = datetime.fromtimestamp(1700000000)   # 从时间戳创建
dt5 = datetime.fromisoformat("2024-01-15T14:30:00")  # ISO字符串解析

print(f"指定时间: {dt1}")
print(f"当前时间: {dt2}")
print(f"时间戳转: {dt4}")
print(f"ISO解析: {dt5}")

# 访问所有属性（继承了date和time的全部属性）
print(f"\n年={dt1.year} 月={dt1.month} 日={dt1.day}")
print(f"时={dt1.hour} 分={dt1.minute} 秒={dt1.second} 微秒={dt1.microsecond}")
print(f"星期: {dt1.strftime('%A')}")  # 英文星期名

# date() 和 time() 方法 —— 拆分
print(f"\n只取日期: {dt1.date()}")
print(f"只取时间: {dt1.time()}")

# 格式化
print(f"\n格式化: {dt1.strftime('%Y-%m-%d %H:%M:%S')}")
print(f"中文: {dt1.strftime('%Y年%m月%d日 %H时%M分%S秒')}")
print(f"ISO: {dt1.isoformat()}")
print(f"字符串解析: {datetime.strptime('2024-01-15 14:30', '%Y-%m-%d %H:%M')}")
# 时间戳转换
print(f"\n转时间戳: {dt1.timestamp()}")
```

### timedelta
- 描述时间段
- 在取值时，只能取到天，秒，微妙。如果需要获取hours，需要自行换算
- 两个datetime相减结果为timedelta类型，两个datetime不可以相加

```python
# ========== 4. timedelta 类（时间段，日期算术核心） ==========

# 创建时间段
td1 = timedelta(days=7)                          # 7天
td2 = timedelta(hours=2, minutes=30)              # 2小时30分
td3 = timedelta(weeks=1, days=3, hours=5)         # 1周3天5小时
td4 = timedelta(days=-1)                          # 负1天（昨天）

print(f"7天: {td1}")
print(f"2小时30分: {td2}")
print(f"1周3天5小时: {td3}")
print(f"-1天: {td4}")

# ⭐ 日期加减运算（最实用！）
now = datetime.now()
print(f"\n当前时间: {now}")
print(f"7天后: {now + td1}")
print(f"7天前: {now - td1}")
print(f"3小时后: {now + td2}")

# ⭐ 两个日期相减 = timedelta
birthday = datetime(2024, 3, 15)
diff = now - birthday
# 2个datetime不能相加
# print(f'datetime add: {(birthday + now())}')
print(f'diff type: {type(diff)}')
print(f"\n距离生日过了: {diff.days} 天")
# 不能获取hours
# print(f"\n距离生日过了: {diff.hours} 小时")
print(f"总秒数: {diff.total_seconds()}")

# timedelta 属性
print(f"\ntd3 的天数: {td3.days}")
print(f"td3 的总秒数: {td3.total_seconds()}")

# 实用：计算年龄
from datetime import date
born = date(2000, 6, 15)
today = date.today()
age = (today - born).days // 365
print(f"\n年龄约: {age} 岁")
```

### 格式化符号速查表

```python
# ========== 5. 格式化符号速查表 ==========
# strftime（对象→字符串） / strptime（字符串→对象）

now = datetime.now()

# 常用格式化符号
print("=== 常用格式化符号 ===")
print(f"%Y 四位年: {now.strftime('%Y')}")
print(f"%m 两位月: {now.strftime('%m')}")
print(f"%d 两位日: {now.strftime('%d')}")
print(f"%H 24小时制: {now.strftime('%H')}")
print(f"%I 12小时制: {now.strftime('%I')}")
print(f"%M 分钟: {now.strftime('%M')}")
print(f"%S 秒: {now.strftime('%S')}")
print(f"%f 微秒: {now.strftime('%f')}")
print(f"%A 星期全名: {now.strftime('%A')}")
print(f"%a 星期缩写: {now.strftime('%a')}")
print(f"%B 月份全名: {now.strftime('%B')}")
print(f"%b 月份缩写: {now.strftime('%b')}")
print(f"%p AM/PM: {now.strftime('%p')}")
print(f"%% 转义%: {now.strftime('100%%')}")

# 解析字符串 → datetime
s = "2024年01月15日 14:30:00"
parsed = datetime.strptime(s, "%Y年%m月%d日 %H:%M:%S")
print(f"\n解析结果: {parsed}")
```

### 时区处理
- timezone(timedelta)定义时区
- datetime的末尾带时区信息

```python
# ========== 6. 时区处理 ==========
from datetime import timezone, timedelta as td

# 创建时区对象
utc = timezone.utc                          # UTC时区
east8 = timezone(td(hours=8))               # 东八区（北京时间）
west5 = timezone(td(hours=-5))              # 西五区（纽约）

# 创建带时区的时间
now_utc = datetime.now(utc)                 # 带UTC时区的当前时间
now_cn = datetime.now(east8)                # 带东八区的当前时间

print(f"UTC时间: {now_utc}")
print(f"北京时间: {now_cn}")

# 时区转换
now_ny = now_utc.astimezone(west5)          # UTC → 纽约
print(f"纽约时间: {now_ny}")

# 去掉时区信息（变为naive datetime）
naive = now_utc.replace(tzinfo=None)
print(f"\n去掉时区: {naive}")

# Python 3.9+ 推荐用 zoneinfo（支持IANA时区名）
# from zoneinfo import ZoneInfo
# beijing = datetime.now(ZoneInfo("Asia/Shanghai"))
# tokyo = beijing.astimezone(ZoneInfo("Asia/Tokyo"))
```

### 时间比较
- datetime带有时区不带时区比较始终为False，只有都指定时区，且时间相等，才会为True

```python
from datetime import datetime

local_datetime = datetime(year=2026, month=8, day=5, hour=10, minute=15, second=0)
utc_datetime = datetime(year=2026, month=8, day=5, hour=2, minute=15, second=0, tzinfo=timezone.utc)
print(f'local_datetime: {local_datetime}')
print(f'utc_datetime: {utc_datetime}')
print(local_datetime == utc_datetime)

dt_utc = datetime(year=2024, month=1, day=15, hour=12, minute=0, second=0, tzinfo=timezone.utc)

dt_cst = datetime(year=2024, month=1, day=15, hour=20, minute=0, second=0, tzinfo=timezone(timedelta(hours=8)))
print(f'dt_utc: {dt_utc}')
print(f'dt_cst: {dt_cst}')
print(dt_utc == dt_cst)
```
